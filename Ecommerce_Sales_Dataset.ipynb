```
import pandas as pd
df=pd.read_csv("Day8_Ecommerce_Sales_Dataset.csv")

print("-------------------- DATA ANALYSIS ----------------")
print("\n\n")
print(" 1. -----Data Selection----- ")
print("\n")
print(df[["Order_ID","Customer","City","Category","Product","Rating"]].head())
print("\n\n")

print(" 2. -----Filtering----- ")
print("\n")
print("Orders where final amount is greater than 1500: ")
high_sale_orders= df[df["Total_Sales"]>1500]
print(high_sale_orders)
print("\n")

print("Orders where Category is Clothing: ")
clothing_orders= df[df["Category"]=="Clothing"]
print("clothing_orders")
print("\n\n")

print(" 3. ------Sorting------- ")
print("Total sales sorted in descending orer is as follows: ")
sorted_sales=df.sort_values("Total_Sales", ascending=False)
print(sorted_sales)
print("\n")
print("Sorted values of Quantity and Ratings are as follows: ")
sorted_categories= df.sort_values(["Quantity", "Rating"], ascending=[False, False])
print(sorted_categories)
print("\n\n")

print(" 4. -------Using groupby()--------")
print("\n")
print("categroy analysis using groupby() and aggregate functions: ")
category_analysis=df.groupby("Category")["Total_Sales"].agg("mean", "max", "min", "sum")
print(category_analysis)
print("\n")
rating_analysis=df.groupby("Category")["Rating"].agg("mean", "max", "min", "sum")
print(rating_analysis)
print("\n\n")

print(" ----------------SALES ANALYSIS--------------")
print("\n\n")
total_sales = df["Total_Sales"].sum()
average_sales = df["Total_Sales"].mean()
highest_sale = df["Total_Sales"].max()
lowest_sale = df["Total_Sales"].min()
total_quantity_sold = df["Quantity"].sum()
no_of_orders = len(df["Order_ID"])

print(f"Total Sales: {total_sales}")
print(f"Average Sales: {average_sales}")
print(f"Highest Sales: {highest_sale}")
print(f"Lowest Sales: {lowest_sale}")
print(f"Total Quantity Sold: {total_quantity_sold}")
print(f"Number of Orders received: {no_of_orders}")
print("\n\n")

print("-------------Top 10 ORDERS BY SALES------------")
print((df[["Order_ID", "Customer", "City", "Category", "Product", "Quantity", "Total_Sales"]].sort_values("Total_Sales", ascending=False)).head(10))

print(" -------------TOP PERFORMING CATEGORY----------- ")
top_category=df.sort_values("Total_Sales", ascending = False)["Category"].iloc[0]
print(f" Top performing category is: {top_category}")

```
